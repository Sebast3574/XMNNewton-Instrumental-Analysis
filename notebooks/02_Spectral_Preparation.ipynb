#### <span style=color:lightgreen> Energy range
To begin, let's first talk about the energy range that I will be looking at. For this purpose I will refer to the count-channel, count-energy graphs (made and showcased in the first notebook), as well as the XMM-Newton user support documentation. I will make it clear what information I am referencing, and if I refer to the user manual, I will clearly indicate page number, section and figure/table number if relevant.

First, let me begin by referring to my count-channel/energy graph in the first notebook. There it can be seen that the line goes vertical at energy values of around $0.2 \ keV$. I have previously interpreted this as a limitation set by the instrument itself. To check this, I examined the user manual and found on page 17 figure 11 (section 3.2.2.1), which documents the effective area of each instrument aboard XMM. The only one relevant for this project being the EPIC:PN. While it is somewhat difficult to see, at $0.2 \ keV$ the effective area is quite low. Furthermore, looking at the quantum efficiencies in figure 28 (p.35, s:3.3.5) one can see that below an energy value of $0.2 \ keV$ the quantum efficiency is lower than the general range. This leads me to believe that the dectector response is reduced and the calibration relability is not up to par. Looking at section 3.3.7.3 it is described about dark currents (sources of background noise) and that these are negligible above $0.25 \ keV$ for the pn camera. The region below this limit will therefore be excluded. However, to give a bit of leeway I will set the lower energy limit to $0.3 \ keV$. 

I can use a similar arguments about data above the $10 \ keV$ range. The effective area (p.17, s:3.2.2.1, fig:11) is quite low, reaching values of $350 \ cm^2$, which is but a mere fraction of the general $800$ to $1200 \ cm^2$ range that most of the instrument is in. This rapid drop above the $10 \ keV$ value means that the source sensitivity is reduced at higher energies. Furthermore from reading pages 43-46 (s:3.3.7.1 and 3.3.7.2) I conclude that at higher energies the background contributes more, reducing signal-to-noise ratio. For these reasons, I will exclude the area above $10 \ keV$. I could also argue that by looking at the counts-energy graph, that this region is mostly dominated by noise and low counts and is therefore not of interest.

To summarize, the chosen energy range will be from $0.3$ to $10 \ keV$ due to limitations of the effective area, background effects, noise and considering the quality of the data.
