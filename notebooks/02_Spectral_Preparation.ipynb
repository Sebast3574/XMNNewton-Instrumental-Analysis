### <span style=color:lightblue> Stage 2: Spectral Preparation
The objective of this stage is to produce a calibrated, response-linked and statistically valid dataset suitable for spectral fitting that will be done in stage 3. I will define the energy range for which it is valid to perform fitting to (because of instrumental properties), verifying the energy redistribution matrix and effective area, validating the PI channel to energy conversion I performed, inspecting and handling background scaling, selecting an appropiate fit statistic and applying a justified method of spectral grouping.
